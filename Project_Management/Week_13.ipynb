# SCRUM (Cont.)
### Product Backlog
- An evolving artifact and items are removed or added at any time
- Adjusted by the product owner
- Regularly reviewed by the product owner
- A prioritized list of features

1. Description
2. Value
3. Estimate
4. Order

### User Stories
- Descriptions of what users want from their perspective
- Three components
    - User
    - Action
    - Benefit
- Epic: A group or collection of user stories
- Acceptance criteria (definition of done) which is usually provided as a checklist to decide if the story is done\
`As a <user> I would like to <want to do thing> so that I can do <action>`

### Story Points

Estimates:
 - Relative Estimates:
    - Not in traditional units of hours, days, or weeks
    - In a relative unit or size

# SCRUM (Cont.)
### Product Backlog
- An evolving artifact and items are removed or added at any time
- Adjusted by the product owner
- Regularly reviewed by the product owner
- A prioritized list of features

1. Description
2. Value
3. Estimate
4. Order

### User Stories
- Descriptions of what users want from their perspective
- Three components
    - User
    - Action
    - Benefit
- Epic: A group or collection of user stories
- Acceptance criteria (definition of done) which is usually provided as a checklist to decide if the story is done\
`As a <user> I would like to <want to do thing> so that I can do <action>`

### Story Points

Estimates:
 - Relative Estimates:
    - Not in traditional units of hours, days, or weeks
    - In a relative unit or size

Use an anchor item or *baseline* to help you determine how risky a story point is.
You can use the story point that seems the easiest or use the medium one to help you set a baseline

#### Sprint (Iterations)
Mini project with planning, execution, deliver, closing, and a retrospective

#### Sprint Planning
Need to gauge team capacity for time and people

Identify items from the sprint backlog

Scrum Master answers:
 - Who is available to work on the sprint
 - What is the average velocity?
 - What can be accomplished during the sprint?
 - What is the sprint backlog and goal?
 - Who is responsible for what tasks?

Other SCRUM Ceremonies:
 - Daily Scrums
 - Sprint Review
 - Sprint Retrospectives

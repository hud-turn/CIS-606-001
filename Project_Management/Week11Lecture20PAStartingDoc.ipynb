2. Suppose you are a project manager in charge of one of the three projects below. Write a project goal using the SMART criteria. \
   • Improve the service of your organization’s IT help desk \
   • Help senior citizens in nursing homes combat privacy risks \
   • Redesign the website of a smart watch company

   
   SMART:
   - Specific: Try to help Sr. citizens fight privacy risks and fraud by providing a few trainings to the these senior citizens.
   - Measurable: The measurable goal is that caretakers, Sr. citizens, or family members of these Sr. citizens report less instances of cyber-fraud against Sr. citizens.
   - Acheivable: A 15% reduction in cyber-fraud is a significant reduction and is likely achievable during a three month extended education campaign
   - Relevant: According to the FBI IC3 center, elder fraud cost people in the United States about $3 billion dollars. As the baby boomer generation reaches retirement age this problem is going to only get worse.
   - Timely: The baby boomer generation is reaching retirement age and this campaign is necessary to help counteract instances of elder and cyber fraud in the community

   Example template:
      - Our goal is to [quantifiable objective] by [timeframe or deadline].
      - [Key players or teams] will accomplish this goal by [what steps you’ll take to achieve the goal]. Accomplishing this goal will [result or benefit].
      - [I or accountable party] will [action word/s] [object of the goal] by [quantity] by [steps] within [time] for the purpose of [relevance/results].
   
      
   SMART Statement:\
   Our goal is to reduce the instances of fraud against senior citizens by 15% in the next year.\
   The city cyber-outreach team will accomplish this goal by helping improve senior cyber-literacy and helping seniors identify scams using a cyberawareness campaign.\
   Accomplishing this goal will reduce the chance that a senior citizen is defrauded.\
   The cyber-outreach team will teach senior citizens how to spot fraud which we hope will reduce instances of cyber-fraud by 15% in the next year.

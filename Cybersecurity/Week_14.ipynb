## Asymmetric Key Encryption
Public and Private key pair, you can use them to establish a conversation and to validate your communications came from you

### Redundancy and failover
Redundancy: Secondary or backup copy of the component, it can take over and help load balance and take over if a node fails (think an HA pair).

Failover: Transfer of workload from a primary system to a secondary system in the event of a failure on the primary system
- Idle standby: B is able to take over A
- Mutual takeover: A backs up for B and vice versa


## Cybersecurity Frameworks

Security Frameworks: Guidelines for building plans to mitigate risks and threats to data and privacy
NIST creates voluntary guidelines that orgs can follow

There are NIST standards such as:
NIST CSF
NIST RMF
NIST Privacy

### NIST CSF 2.0

Core: Govern, Identify, Protect, Detect, Respond, and Recover

Profiles: Identify opportunites for improving cybersecurity posture by comparing a current with target profile

Tiers: Looks at risk apetite and the threat profile of the organization, talks about what steps your org and department should be looking at implementing (in terms of risk management)

### Vulnerabilities

N-days or known vulns: Where there is a vuln on a system and it was dropped a while ago
0 Days: Vulns that are unknown and are out there in the wild

Vuln Scanning
1. Define the scope of a vuln scan
2. Scanning automatically scan a system for known vulns
3. Classification and Evaluation report vuln and sort by severity
4. Remediation determine which vulns to mitigate and how
5. Rescanning Verify the effectiveness of remediation

### Penetration Testing

Authorized security expert that tries to stress test the system.
Attacks vs penetration testing is authorized vs unauthorized.

N-day exploits: A exploit is made known to the authors of the affected sotare, the vuln is often fixed through a patch to make the exploit unusable
0-day exploits: Exploits unknown to mostly everyone but people that developed them are the most dangerous exploits.

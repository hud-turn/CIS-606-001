# Integrity
Data can be covered by data backups
Differential, incremental, and full

Pre-scheduled and need to be notified about disruptions
Some services can't be shut down

It is hard to anticpate problems or failures and stuff happens (look at cloudflare for example).

How to choose:
Volatility: Data that changes frequently
Importance: If that data is core to your business
Difficulty: How much of a headache it will be to backup and size of the DB.

Cold Backup/offline backup: Database is shut down
Hot Backup: is running and online
Full Backup: Backup the entire DB
Incremental Backup: The only data backed up here is only the data that is changed since the last backup
Differential Backups: Only copy data changed since the last full backup
